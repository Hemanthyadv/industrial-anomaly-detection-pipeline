# Industrial IoT Real-Time Platform: End-to-End Walkthrough

This notebook walks through the complete ML lifecycle for industrial sensor intelligence:
1. **Data Generation & Loading**: Generate historical multi-sensor dataset with injected synthetic anomalies
2. **Exploratory Data Analysis**: Inspect sensor distributions, types, and sites
3. **Visualizing Normal Sensor Dynamics**: Seasonality, baselines, and physical noise
4. **Visualizing Injected Anomalies**: Spikes, drifts, sensor freezes, and noise bursts
5. **Feature Engineering**: Multi-variate alignment and rolling statistical windows
6. **Model Training**: Isolation Forest, Local Outlier Factor (LOF), and Neural Autoencoder
7. **Model Comparison**: Cross-model performance comparison (Precision, Recall, F1, FPR, ROC-AUC)
8. **Evaluation & Confusion Matrices**: Detailed classification performance breakdown
9. **Time-Series Forecasting**: Baseline vs. Prophet models for proactive anomaly prevention
10. **Forecast Confidence Intervals**: Visualizing prediction intervals and anomaly thresholds
11. **Real-Time Inference Simulation**: Simulating streaming predictions against API payload format

```python
# Setup environment and sys.path
import os, sys
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set plotting defaults
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (14, 5)
plt.rcParams['font.size'] = 11
print("Environment initialized successfully.")
```

## Step 1: Generate & Load 7-Day Sensor Dataset

```python
from src.kafka.sensor_simulator import generate

data_path = os.path.join('..', 'data', 'sample_sensor_data.csv')
# Generate 7-day dataset with 120 sensors if not already generated
total_rows = generate(days=7, sensors=120, output=data_path, seed=42)
df = pd.read_csv(data_path)
print(f"Loaded {len(df):,} sensor observations across {df['sensor_id'].nunique()} unique sensors.")
df.head()
```

## Step 2: Exploratory Data Analysis

```python
print("--- Dataset Summary ---")
print(f"Date Range: {df['timestamp'].min()} to {df['timestamp'].max()}")
print(f"Sensor Types: {df['sensor_type'].value_counts().to_dict()}")
print(f"Sites: {df['site'].value_counts().to_dict()}")
print(f"Total Injected Anomalies: {df['is_anomaly'].sum():,} ({df['is_anomaly'].mean()*100:.2f}%)")
print(f"Anomaly Breakdown:\n{df[df['is_anomaly'] == 1]['anomaly_type'].value_counts()}")
```

## Step 3 & 4: Visualizing Sensor Behavior & Injected Faults

```python
# Visualize single temperature sensor over 48 hours
s0 = df[df['sensor_id'] == 'sensor-0000'].head(2880).copy()
s0['dt'] = pd.to_datetime(s0['timestamp'])

plt.figure(figsize=(15, 6))
plt.plot(s0['dt'], s0['value'], label='Sensor-0000 Reading (Normal + Injected)', color='#1f77b4', lw=1.2)
anomalies = s0[s0['is_anomaly'] == 1]
if not anomalies.empty:
    plt.scatter(anomalies['dt'], anomalies['value'], color='#d62728', s=60, zorder=5, label='Injected Anomaly')

plt.title('Temperature Sensor Profile (Normal Diurnal Cycles & Injected Spikes)', fontsize=13)
plt.xlabel('Timestamp')
plt.ylabel('Temperature (°C)')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## Step 5: Feature Engineering Matrix

```python
from src.ml.train_models import load_training_data

# Prepare aligned multi-sensor feature matrix (Temperature, Pressure, Vibration)
X, y, feature_df = load_training_data(data_path, sample_frac=1.0, seed=42)
print(f"Engineered Feature Matrix X: {X.shape}, Labels y: {y.shape} (Anomalies: {y.sum():,})")

feature_summary = pd.DataFrame(X, columns=['Temperature', 'Pressure', 'Vibration']).describe()
feature_summary
```

## Step 6, 7 & 8: Train Models, Compare Performance & Evaluate

```python
from src.ml.train_models import train_isolation_forest, train_lof, train_autoencoder
from src.ml.model_evaluation import evaluate_anomaly

# Train/Test split (80/20 temporal split)
split = int(len(X) * 0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

print(f"Training set: {len(X_train):,} rows | Test set: {len(X_test):,} rows")

# 1. Isolation Forest
if_model, if_train_metrics = train_isolation_forest(X_train, y_train, n_estimators=200, contamination=0.02)
if_preds = (if_model.predict(X_test) == -1).astype(int)
if_scores = -if_model.decision_function(X_test)
if_eval = evaluate_anomaly(y_test, if_preds, if_scores)

# 2. Local Outlier Factor
lof_model, lof_train_metrics = train_lof(X_train, y_train, n_neighbors=20, contamination=0.02)
lof_preds = (lof_model.predict(X_test) == -1).astype(int)
lof_scores = -lof_model.decision_function(X_test)
lof_eval = evaluate_anomaly(y_test, lof_preds, lof_scores)

# 3. Autoencoder (MLP-based)
ae_bundle, ae_train_metrics = train_autoencoder(X_train, y_train, max_iter=150)
ae_scaled = ae_bundle['scaler'].transform(X_test)
ae_recon = ae_bundle['mlp'].predict(ae_scaled)
ae_errors = np.mean((ae_scaled - ae_recon) ** 2, axis=1)
ae_preds = (ae_errors > ae_bundle['threshold']).astype(int)
ae_eval = evaluate_anomaly(y_test, ae_preds, ae_errors)

# Comparison Table
comparison = pd.DataFrame([
    {"Model": "Isolation Forest", "Precision": if_eval["precision"], "Recall": if_eval["recall"], "F1": if_eval["f1"], "ROC-AUC": if_eval["roc_auc"], "FPR": if_eval["false_positive_rate"]},
    {"Model": "Local Outlier Factor", "Precision": lof_eval["precision"], "Recall": lof_eval["recall"], "F1": lof_eval["f1"], "ROC-AUC": lof_eval["roc_auc"], "FPR": lof_eval["false_positive_rate"]},
    {"Model": "Autoencoder (MLP)", "Precision": ae_eval["precision"], "Recall": ae_eval["recall"], "F1": ae_eval["f1"], "ROC-AUC": ae_eval["roc_auc"], "FPR": ae_eval["false_positive_rate"]},
])
print("=== Model Evaluation on Test Set ===")
comparison.round(4)
```

## Step 9 & 10: Time-Series Forecasting & Prediction Intervals

```python
from src.ml.forecasting import train_forecast_model

# Train and evaluate Prophet forecasting model for representative sensor
forecast_result = train_forecast_model(df, sensor_id='sensor-0000', train_fraction=0.85)

if 'error' not in forecast_result:
    metrics = forecast_result['metrics']
    base_metrics = forecast_result['baseline_metrics']
    print(f"Prophet Forecast -> MAE: {metrics['mae']:.4f}, RMSE: {metrics['rmse']:.4f}")
    print(f"Naive Baseline   -> MAE: {base_metrics['mae']:.4f}, RMSE: {base_metrics['rmse']:.4f}")
    
    # Plot forecast with confidence intervals
    forecast_df = forecast_result['forecast'].tail(1440) # Last 24 hours
    plt.figure(figsize=(15, 6))
    plt.plot(forecast_df['ds'], forecast_df['yhat'], label='Point Forecast ($\\hat{y}$)', color='#2ca02c')
    plt.fill_between(forecast_df['ds'], forecast_df['yhat_lower'], forecast_df['yhat_upper'], 
                     alpha=0.3, color='#2ca02c', label='95% Prediction Interval')
    plt.title('24-Hour Ahead Temperature Forecast with Confidence Intervals', fontsize=13)
    plt.xlabel('Time')
    plt.ylabel('Temperature (°C)')
    plt.legend(loc='upper right')
    plt.tight_layout()
    plt.show()
else:
    print(f"Forecasting note: {forecast_result['error']}")
```
